```python
import torch
import numpy as np

device = torch.device('cuda:0' if torch.cuda.is_available() else 'cpu')
print(device)
```

### Используем модель на GPT-2 для диалога

```python
from transformers import GPT2LMHeadModel, GPT2TokenizerFast

checkpoint = 'microsoft/DialoGPT-medium'
tokenizer = GPT2TokenizerFast.from_pretrained(checkpoint)
model = GPT2LMHeadModel.from_pretrained(checkpoint)

model = model.to(device)
```

```python
model = model.eval()
```

```python
def get_length_params(text:str, tokenizer) -> str:
    tokens_count = len(tokenizer.encode(text))

    if tokens_count <= 15:
        return '1'
    if tokens_count <= 50:
        return '2'
    if tokens_count <= 256:
        return '3'
    return '-'
```

### Общение с моделью

```python
@torch.inference_mode()
def chat():
    chat_history_ids = torch.zeros((1, 0), dtype = torch.int).to(device)

    human = True

    try:
        while True:
            if human:
                input_user = input('-> Человек: ')
                new_user_input_ids = tokenizer.encode(
                    f"|0|{get_length_params(input_user, tokenizer)}|" + input_user + tokenizer.eos_token,
                    return_tensors = 'pt'
                ).to(device)
                chat_history_ids = torch.cat([chat_history_ids, new_user_input_ids], dim = -1)
            else:
                new_user_input_ids = tokenizer.encode(f"|1|{np.random.choice(['2', '3'])}|", return_tensors = 'pt').to(device)
                chat_history_ids = torch.cat([chat_history_ids, new_user_input_ids], dim = -1)

                input_len = chat_history_ids.shape[-1]

                chat_history_ids = model.generate(
                    chat_history_ids,
                    num_return_sequences = 1,
                    max_new_tokens = 256,
                    no_repeat_ngram_size = 2, # чтобы не повторялись n-граммы размера 2
                    do_sample = True, # не сэмплирпованная генерация
                    top_k = 50, # выбираем из топ-50 слов
                    top_p = 0.9, # выбираем слова, чтобы суммарная вероятность была >= 90 процентов
                    temperature = 0.5, # насколько большой разброс смысла
                    eos_token_id = tokenizer.eos_token_id,
                    pad_token_id = tokenizer.pad_token_id,
                )
                print(f'-> GPT: {tokenizer.decode(chat_history_ids[:, input_len:][0], skip_special_tokens = True)}')
                if chat_history_ids.shape[-1] > 600:
                    return
            human = not human
    except KeyboardInterrupt:
        return
```

```python
chat()
```
